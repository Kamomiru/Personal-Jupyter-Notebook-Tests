#### Now to the fun stuff: MATH!

These equations are all written in LaTeX. A programming language to create scientific papers and documentation.
You can find more about it here: [LaTeX Mathematical Expressions](https://www.overleaf.com/learn/latex/Mathematical_expressions)

---

We can either display simple inline equations using `$Math$` with one `$` sign: $E = mc^2$ so we can continue writing after.

Or display *the scariest equation ever* as a "standalone" using two `$$`:

$$
a^2 + b^2 = c^2
$$

---

We can add things like **fractions**, **lowercases** or **special characters** and many more!
These get inserted by using backslash `\` followed by a keyword `\keyword`

Special letter `\alpha` and fraction `\frac{}{}`:
$$sin(\beta) = \frac{b}{c}$$

single lowercase letters with `_`:
$$E = E_P + E_K$$

or multiple letters with `_{}`
$$E = E_{pot} + E_{kin}$$
